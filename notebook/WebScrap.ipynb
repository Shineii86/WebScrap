```python
#@title 🔧 Step 1 — Setup { display-mode: "form" }

#@markdown Install Chromium, Selenium, and dependencies. Takes ~60s on first run.

#@markdown ---

import os, sys, subprocess, shutil, warnings, json

# ── Suppress warnings ──
warnings.filterwarnings("ignore", category=UserWarning)
os.environ["TF_CPP_MIN_LOG_LEVEL"] = "3"

# ── Clone repo for src/ package ──
REPO_DIR = "/content/WebScrap"
if not os.path.isdir(os.path.join(REPO_DIR, "src")):
    subprocess.run(["git", "clone", "-q", "https://github.com/Shineii86/WebScrap.git", REPO_DIR],
                   capture_output=True)

if REPO_DIR not in sys.path:
    sys.path.insert(0, REPO_DIR)

# ══════════════════════════════════════════════════════
#  SETUP
# ══════════════════════════════════════════════════════

# ── Install system packages ──
print("🔹 Installing system packages...")
subprocess.run(["apt-get", "-y", "-qq", "update"], capture_output=True)
subprocess.run(["apt-get", "-y", "-qq", "install", "-qq",
    "chromium-browser", "chromium-chromedriver",
    "xvfb", "fonts-liberation", "libnss3", "libatk-bridge2.0-0",
    "libdrm2", "libxcomposite1", "libxdamage1", "libxrandr2",
    "libgbm1", "libasound2", "libpangocairo-1.0-0", "libgtk-3-0"],
    capture_output=True)
print("✅ System packages installed")

# ── Install Python packages (BEFORE importing src/) ──
print("🔹 Installing Python packages...")
subprocess.run([sys.executable, "-m", "pip", "install", "-q",
    "selenium", "beautifulsoup4", "requests", "lxml", "webdriver-manager"],
    capture_output=True)
print("✅ Python packages installed")

# ── NOW import shared UI (selenium is installed) ──
from src.ui import show_header, show_ok, show_warn, show_err, show_info, show_step

show_header("🔧", "Setup", "Installing Chromium, Selenium & dependencies")

# ── Detect Chromium binary ──
show_step("Detecting Chromium...")

CHROME_BIN = None
for path in [
    "/usr/bin/chromium-browser",
    "/usr/bin/chromium",
    "/usr/bin/google-chrome",
    "/usr/bin/google-chrome-stable",
    shutil.which("chromium-browser"),
    shutil.which("chromium"),
    shutil.which("google-chrome"),
]:
    if path and os.path.isfile(path):
        CHROME_BIN = path
        break

if CHROME_BIN:
    result = subprocess.run([CHROME_BIN, "--version"], capture_output=True, text=True)
    show_ok(f"Chromium: {result.stdout.strip() if result.returncode == 0 else CHROME_BIN}")
else:
    show_err("Chromium not found — try restarting runtime")

# ── Auto-download matching ChromeDriver via webdriver-manager ──
show_step("Downloading matching ChromeDriver...")

CHROMEDRIVER_BIN = None

# Strategy 1: webdriver-manager (downloads correct version)
try:
    from webdriver_manager.chrome import ChromeDriverManager
    from webdriver_manager.core.os_manager import ChromeType
    CHROMEDRIVER_BIN = ChromeDriverManager(chrome_type=ChromeType.CHROMIUM).install()
    show_ok(f"ChromeDriver (auto): {CHROMEDRIVER_BIN}")
except Exception as e:
    show_warn(f"webdriver-manager failed: {str(e)[:100]}")

# Strategy 2: system chromedriver fallback
if not CHROMEDRIVER_BIN:
    show_step("Trying system chromedriver...")
    for path in [
        "/usr/lib/chromium/chromedriver",
        "/usr/lib/chromium-browser/chromedriver",
        "/usr/bin/chromedriver",
        shutil.which("chromedriver"),
    ]:
        if path and os.path.isfile(path):
            CHROMEDRIVER_BIN = path
            break
    if CHROMEDRIVER_BIN:
        show_ok(f"ChromeDriver (system): {CHROMEDRIVER_BIN}")

# Save config for Step 2
DRIVER_CONFIG = {"chrome_bin": CHROME_BIN, "chromedriver_bin": CHROMEDRIVER_BIN}
with open("/tmp/webscrap_driver_config.json", "w") as f:
    json.dump(DRIVER_CONFIG, f)

if not CHROMEDRIVER_BIN:
    show_err("ChromeDriver not found — try restarting runtime")

# ── Quick Selenium test ──
if CHROME_BIN and CHROMEDRIVER_BIN:
    show_step("Verifying Selenium connection...")
    try:
        from selenium import webdriver
        from selenium.webdriver.chrome.options import Options
        from selenium.webdriver.chrome.service import Service
        test_opts = Options()
        test_opts.binary_location = CHROME_BIN
        test_opts.add_argument("--headless=new")
        test_opts.add_argument("--no-sandbox")
        test_opts.add_argument("--disable-dev-shm-usage")
        test_service = Service(executable_path=CHROMEDRIVER_BIN)
        test_driver = webdriver.Chrome(service=test_service, options=test_opts)
        test_driver.get("data:text/html,<h1>OK</h1>")
        assert "OK" in test_driver.page_source
        test_driver.quit()
        show_ok("Selenium connection verified")
    except Exception as e:
        show_err(f"Selenium test failed: {str(e)[:200]}")

# ── Create output directory ──
OUTPUT_DIR = "/content/webscrap_output"
os.makedirs(OUTPUT_DIR, exist_ok=True)
show_ok(f"Output directory: {OUTPUT_DIR}")

show_ok("Setup complete!", "🎉")
show_info("Run the next cell to scrape a website", "➡️")
```

```python
#@title 🕷️ Step 2 — Scrape Website { display-mode: "form" }

#@markdown ### Enter URL
#@markdown Paste any website URL below. Works with SPAs, Next.js, React, Vue, Angular, and static sites.

url = "" #@param {type:"string"}

#@markdown ### Settings
#@markdown **Timeout** — max seconds to wait for the page to load  
#@markdown **Wait After Load** — extra seconds for JavaScript to settle (increase for heavy SPAs)  
#@markdown **Auto-Scroll** — scroll to bottom to trigger lazy-loaded content  
#@markdown **Custom JS** — optional JavaScript to execute after page loads (e.g., click buttons, expand menus)

timeout = 30 #@param {type:"slider", min:10, max:120, step:5}
wait_after_load = 3 #@param {type:"slider", min:0, max:30, step:1}
auto_scroll = True #@param {type:"boolean"}
custom_js = "" #@param {type:"string"}

#@markdown ### Output
#@markdown **Save HTML** — download the captured HTML as a file  
#@markdown **Show Preview** — display a syntax-highlighted preview  
#@markdown **Show Links** — extract and display all links found

save_html_file = True #@param {type:"boolean"}
show_preview = True #@param {type:"boolean"}
show_links = True #@param {type:"boolean"}

#@markdown ---

import os, sys, time, hashlib, json, html as html_lib
from datetime import datetime, timezone
from urllib.parse import urlparse, urljoin

# ── Ensure src/ is importable ──
REPO_DIR = "/content/WebScrap"
if REPO_DIR not in sys.path:
    sys.path.insert(0, REPO_DIR)

from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.support.ui import WebDriverWait
from bs4 import BeautifulSoup
from src.ui import (
    show_header, show_ok, show_warn, show_err, show_info, show_step,
    show_stats, show_html_preview, show_links_table, show_meta_tags,
)
from src.config import USER_AGENT

# ── Load driver config from Step 1 ──
CHROME_BIN = None
CHROMEDRIVER_BIN = None

try:
    with open("/tmp/webscrap_driver_config.json") as f:
        _cfg = json.load(f)
    CHROME_BIN = _cfg.get("chrome_bin")
    CHROMEDRIVER_BIN = _cfg.get("chromedriver_bin")
except FileNotFoundError:
    pass

# Fallback: re-detect Chromium
if not CHROME_BIN:
    import shutil
    for p in ["/usr/bin/chromium-browser", "/usr/bin/chromium", "/usr/bin/google-chrome", shutil.which("chromium-browser"), shutil.which("chromium")]:
        if p and os.path.isfile(p):
            CHROME_BIN = p
            break

# Fallback: re-detect ChromeDriver (system or webdriver-manager)
if not CHROMEDRIVER_BIN:
    import shutil
    for p in ["/usr/lib/chromium/chromedriver", "/usr/lib/chromium-browser/chromedriver", "/usr/bin/chromedriver", shutil.which("chromedriver")]:
        if p and os.path.isfile(p):
            CHROMEDRIVER_BIN = p
            break

# Fallback: webdriver-manager auto-download
if not CHROMEDRIVER_BIN:
    try:
        from webdriver_manager.chrome import ChromeDriverManager
        from webdriver_manager.core.os_manager import ChromeType
        CHROMEDRIVER_BIN = ChromeDriverManager(chrome_type=ChromeType.CHROMIUM).install()
    except Exception:
        pass

if not CHROME_BIN:
    show_err("Chromium not found. Run Step 1 first.")
    raise SystemExit("Chromium not found")

if not CHROMEDRIVER_BIN:
    show_err("ChromeDriver not found. Run Step 1 first.")
    raise SystemExit("ChromeDriver not found")

# ── Validate URL ──
if not url or not url.strip():
    show_err("Please enter a URL in the field above")
    raise SystemExit("No URL provided")

url = url.strip()
if not url.startswith(('http://', 'https://')):
    url = 'https://' + url

parsed = urlparse(url)
if not parsed.netloc:
    show_err(f"Invalid URL: {url}")
    raise SystemExit("Invalid URL")

show_header("🕷️", "Scraping", url)

# ── Build driver ──
show_step("Launching headless Chromium...")

chrome_opts = Options()
chrome_opts.binary_location = CHROME_BIN
for arg in [
    "--headless=new", "--no-sandbox", "--disable-dev-shm-usage",
    "--disable-gpu", "--disable-extensions", "--disable-infobars",
    "--window-size=1920,1080",
    f"--user-agent={USER_AGENT}",
]:
    chrome_opts.add_argument(arg)

service = Service(executable_path=CHROMEDRIVER_BIN)
driver = webdriver.Chrome(service=service, options=chrome_opts)
driver.set_page_load_timeout(timeout)

show_ok("Chromium launched")

# ── Navigate ──
show_step(f"Loading page (timeout: {timeout}s)...")
start_time = time.time()

try:
    driver.get(url)

    WebDriverWait(driver, timeout).until(
        lambda d: d.execute_script("return document.readyState") == "complete"
    )
    show_ok(f"Page loaded ({round(time.time() - start_time, 1)}s)")

    if wait_after_load > 0:
        show_step(f"Waiting {wait_after_load}s for JS to settle...")
        time.sleep(wait_after_load)

    # Auto-scroll
    scroll_count = 0
    if auto_scroll:
        show_step("Scrolling to load lazy content...")
        last_h = driver.execute_script("return document.body.scrollHeight")
        for _ in range(50):
            driver.execute_script("window.scrollTo(0, document.body.scrollHeight)")
            time.sleep(0.8)
            new_h = driver.execute_script("return document.body.scrollHeight")
            scroll_count += 1
            if new_h == last_h:
                break
            last_h = new_h
        driver.execute_script("window.scrollTo(0, 0)")
        show_ok(f"Scrolled {scroll_count} times")

    # Custom JS
    if custom_js and custom_js.strip():
        show_step("Executing custom JavaScript...")
        try:
            driver.execute_script(custom_js)
            time.sleep(1)
            show_ok("Custom JS executed")
        except Exception as js_err:
            show_warn(f"Custom JS error: {str(js_err)[:100]}")

    # ── Capture HTML ──
    show_step("Capturing rendered HTML...")
    html = driver.page_source
    final_url = driver.current_url
    title = driver.title
    load_time = round(time.time() - start_time, 2)

    show_ok(f"Captured {len(html):,} characters ({len(html.encode('utf-8')) / 1024:.1f} KB)")

except Exception as e:
    show_err(f"Failed to load page: {str(e)[:200]}")
    driver.quit()
    raise

finally:
    driver.quit()

# ── Parse metadata ──
show_step("Extracting metadata...")
soup = BeautifulSoup(html, "html.parser")

meta_tags = {}
for tag in soup.find_all("meta"):
    name = tag.get("name") or tag.get("property") or tag.get("http-equiv")
    content = tag.get("content")
    if name and content:
        meta_tags[name] = content

links = []
for a in soup.find_all("a", href=True):
    href = a["href"]
    if not href.startswith(('http://', 'https://', 'mailto:', 'tel:', '#')):
        href = urljoin(final_url, href)
    links.append({"text": a.get_text(strip=True)[:120], "url": href})

images = []
for img in soup.find_all("img"):
    src = img.get("src", "")
    if src and not src.startswith(('data:', 'http://', 'https://')):
        src = urljoin(final_url, src)
    images.append({"src": src, "alt": img.get("alt", "")})

scripts = len(soup.find_all("script"))
stylesheets = len(soup.find_all("link", rel="stylesheet"))
content_hash = hashlib.sha256(html.encode("utf-8")).hexdigest()[:16]

show_ok(f"Found {len(links)} links, {len(images)} images, {scripts} scripts, {stylesheets} stylesheets")

# ── Stats (via shared UI) ──
show_stats([
    ("📄", "Characters", f"{len(html):,}"),
    ("📦", "Size", f"{len(html.encode('utf-8')) / 1024:.1f} KB"),
    ("🔗", "Links", str(len(links))),
    ("🖼️", "Images", str(len(images))),
    ("⏱️", "Load Time", f"{load_time}s"),
    ("📜", "Scrolls", str(scroll_count)),
])

# ── Redirect notice ──
if final_url != url:
    show_info(f"Redirected: {url} → {final_url}", "↪️")

# ── Title & Description ──
if title:
    show_ok(f"Title: {title}", "🏷️")

desc = meta_tags.get("og:description") or meta_tags.get("description") or meta_tags.get("twitter:description")
if desc:
    show_info(f"Description: {desc[:200]}")

# ── Save HTML ──
if save_html_file:
    domain = parsed.netloc.replace("www.", "")
    safe_name = "".join(c if c.isalnum() or c in ('-', '.') else '_' for c in domain)
    ts = datetime.now().strftime("%Y%m%d_%H%M%S")
    filename = f"{safe_name}_{ts}.html"
    filepath = os.path.join(OUTPUT_DIR, filename)
    with open(filepath, 'w', encoding='utf-8') as f:
        f.write(html)
    show_ok(f"Saved: {filepath}", "💾")

    meta_path = filepath.replace('.html', '_meta.json')
    meta_data = {
        "url": url,
        "final_url": final_url,
        "title": title,
        "char_count": len(html),
        "file_size_bytes": len(html.encode('utf-8')),
        "link_count": len(links),
        "image_count": len(images),
        "script_count": scripts,
        "stylesheet_count": stylesheets,
        "scroll_count": scroll_count,
        "load_time_s": load_time,
        "content_hash": content_hash,
        "meta_tags": meta_tags,
        "scraped_at": datetime.now(timezone.utc).isoformat(),
    }
    with open(meta_path, 'w', encoding='utf-8') as f:
        json.dump(meta_data, f, indent=2, ensure_ascii=False)
    show_ok(f"Metadata: {meta_path}", "📋")

# ── HTML Preview (via shared UI) ──
if show_preview:
    show_html_preview(html)

# ── Links Table (via shared UI) ──
if show_links:
    show_links_table(links)

# ── Meta Tags (via shared UI) ──
show_meta_tags(meta_tags)

show_header("✅", "Done!", f"Scraped {title or url} in {load_time}s")
show_info("Run this cell again with a different URL to scrape more pages", "🔄")
```

```python
#@title 💾 Step 3 — Download & Export { display-mode: "form" }

#@markdown Download all scraped files as a ZIP archive.

#@markdown ---

import os, sys, glob, zipfile
from datetime import datetime

# ── Ensure src/ is importable ──
REPO_DIR = "/content/WebScrap"
if REPO_DIR not in sys.path:
    sys.path.insert(0, REPO_DIR)

from src.ui import show_header, show_ok, show_warn, show_err, show_info, show_step

OUTPUT_DIR = "/content/webscrap_output"

show_header("💾", "Export", "Download scraped files")

# ── Find files ──
html_files = sorted(glob.glob(os.path.join(OUTPUT_DIR, "*.html")))
json_files = sorted(glob.glob(os.path.join(OUTPUT_DIR, "*.json")))
all_files = html_files + json_files

if not all_files:
    show_warn("No files found. Run Step 2 first to scrape a website.")
else:
    show_info(f"Found {len(html_files)} HTML file(s) and {len(json_files)} metadata file(s)")

    # ── Create ZIP ──
    ts = datetime.now().strftime("%Y%m%d_%H%M%S")
    zip_name = f"webscrap_export_{ts}.zip"
    zip_path = os.path.join("/content", zip_name)

    with zipfile.ZipFile(zip_path, 'w', zipfile.ZIP_DEFLATED) as zf:
        for fpath in all_files:
            zf.write(fpath, os.path.basename(fpath))

    zip_size = os.path.getsize(zip_path) / 1024
    show_ok(f"Created {zip_name} ({zip_size:.1f} KB)")

    for f in all_files:
        size = os.path.getsize(f) / 1024
        show_info(f"  📄 {os.path.basename(f)} ({size:.1f} KB)")

    # ── Download ──
    try:
        from google.colab import files as colab_files
        show_step("Triggering download...")
        colab_files.download(zip_path)
        show_ok(f"Downloaded {zip_name}")
    except ImportError:
        show_info(f"Not on Colab. File saved at: {zip_path}")
    except Exception as e:
        show_warn(f"Download failed: {str(e)[:100]}")
        show_info(f"File is at: {zip_path}")

show_header("✅", "All Done!", "Your scraped data is ready")
```
