```python
#@title 🔧 Step 1 — Setup { display-mode: "form" }

#@markdown Install Chromium, Selenium, and dependencies. Takes ~60s on first run.

#@markdown ---

import os, sys, subprocess, warnings
from IPython.display import display, HTML

# ── Theme-safe UI tokens ──
BG_CARD  = '#f4f5f7'
BG_INNER = '#ffffff'
BORDER   = '#dde1e8'
TEXT     = '#2c2c3a'
TEXT_DIM = '#6b7280'
TEXT_MUTED = '#9ca3af'
FONT = "font-family:system-ui,-apple-system,'Segoe UI',Roboto,sans-serif"
MONO = "font-family:'SF Mono',SFMono-Regular,Consolas,monospace"
C_OK   = ('#16a34a', '#f0fdf4', '#15803d')
C_WARN = ('#d97706', '#fffbeb', '#92400e')
C_ERR  = ('#dc2626', '#fef2f2', '#991b1b')
C_INFO = ('#2563eb', '#eff6ff', '#1e40af')
C_STEP = ('#7c3aed', '#f5f3ff', '#5b21b6')

def _card(inner):
    return f'<div style="background:{BG_CARD};border:1px solid {BORDER};border-radius:16px;padding:2px;margin:14px 0"><div style="background:{BG_INNER};border-radius:14px;padding:24px">{inner}</div></div>'

def _header(icon, title, sub=""):
    sub_h = f'<div style="{FONT};font-size:12.5px;color:{TEXT_DIM};margin-top:2px">{sub}</div>' if sub else ''
    return f'<div style="display:flex;align-items:center;gap:12px;margin-bottom:20px;padding-bottom:14px;border-bottom:2px solid {BORDER}"><div style="width:44px;height:44px;display:flex;align-items:center;justify-content:center;background:linear-gradient(135deg,#667eea,#764ba2);border-radius:12px;font-size:22px">{icon}</div><div><div style="{FONT};font-size:20px;font-weight:700;color:{TEXT}">{title}</div>{sub_h}</div></div>'

def _row(msg, colors, icon=""):
    b, bg, t = colors
    return f'<div style="display:flex;align-items:center;gap:10px;padding:10px 14px;margin:6px 0;border-radius:10px;border-left:3px solid {b};background:{bg};color:{t};{MONO};font-size:13.5px">{icon} {msg}</div>'

def ce_show(icon, title, sub=""):
    display(HTML(_card(_header(icon, title, sub))))
def ce_ok(msg, icon="✅"):
    display(HTML(_row(msg, C_OK, icon)))
def ce_warn(msg, icon="⚠️"):
    display(HTML(_row(msg, C_WARN, icon)))
def ce_err(msg, icon="❌"):
    display(HTML(_row(msg, C_ERR, icon)))
def ce_info(msg, icon="ℹ️"):
    display(HTML(_row(msg, C_INFO, icon)))
def ce_step(msg, icon="🔹"):
    display(HTML(_row(msg, C_STEP, icon)))

# ── Suppress warnings ──
warnings.filterwarnings("ignore", category=UserWarning)
os.environ["TF_CPP_MIN_LOG_LEVEL"] = "3"

# ══════════════════════════════════════════════════════
#  SETUP
# ══════════════════════════════════════════════════════

ce_show("🔧", "Setup", "Installing Chromium, Selenium & dependencies")

# ── Install system packages ──
ce_step("Installing system packages...")
subprocess.run(["apt-get", "-y", "-qq", "update"], capture_output=True)
subprocess.run(["apt-get", "-y", "-qq", "install", "-qq",
    "chromium-browser", "chromium-chromedriver",
    "xvfb", "fonts-liberation", "libnss3", "libatk-bridge2.0-0",
    "libdrm2", "libxcomposite1", "libxdamage1", "libxrandr2",
    "libgbm1", "libasound2", "libpangocairo-1.0-0", "libgtk-3-0"],
    capture_output=True)
ce_ok("System packages installed")

# ── Install Python packages ──
ce_step("Installing Python packages...")
subprocess.run([sys.executable, "-m", "pip", "install", "-q",
    "selenium", "beautifulsoup4", "requests", "lxml"],
    capture_output=True)
ce_ok("Python packages installed")

# ── Verify Chromium ──
ce_step("Verifying Chromium...")
result = subprocess.run(["chromium-browser", "--version"], capture_output=True, text=True)
if result.returncode == 0:
    ce_ok(f"Chromium: {result.stdout.strip()}")
else:
    ce_err("Chromium not found — try restarting runtime")

# ── Create output directory ──
OUTPUT_DIR = "/content/webscrap_output"
os.makedirs(OUTPUT_DIR, exist_ok=True)
ce_ok(f"Output directory: {OUTPUT_DIR}")

ce_ok("Setup complete!", "🎉")
ce_info("Run the next cell to scrape a website", "➡️")
```

```python
#@title 🕷️ Step 2 — Scrape Website { display-mode: "form" }

#@markdown ### Enter URL
#@markdown Paste any website URL below. Works with SPAs, Next.js, React, Vue, Angular, and static sites.

url = "" #@param {type:"string"}

#@markdown ### Settings
#@markdown **Timeout** — max seconds to wait for the page to load  
#@markdown **Wait After Load** — extra seconds for JavaScript to settle (increase for heavy SPAs)  
#@markdown **Auto-Scroll** — scroll to bottom to trigger lazy-loaded content  
#@markdown **Custom JS** — optional JavaScript to execute after page loads (e.g., click buttons, expand menus)

timeout = 30 #@param {type:"slider", min:10, max:120, step:5}
wait_after_load = 3 #@param {type:"slider", min:0, max:30, step:1}
auto_scroll = True #@param {type:"boolean"}
custom_js = "" #@param {type:"string"}

#@markdown ### Output
#@markdown **Save HTML** — download the captured HTML as a file  
#@markdown **Show Preview** — display a syntax-highlighted preview  
#@markdown **Show Links** — extract and display all links found

save_html_file = True #@param {type:"boolean"}
show_preview = True #@param {type:"boolean"}
show_links = True #@param {type:"boolean"}

#@markdown ---

import os, sys, time, hashlib, json
from datetime import datetime, timezone
from urllib.parse import urlparse, urljoin

from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.support.ui import WebDriverWait
from bs4 import BeautifulSoup

from IPython.display import display, HTML

# ── UI helpers (re-declare if needed) ──
try:
    ce_show
except NameError:
    BG_CARD='#f4f5f7'; BG_INNER='#ffffff'; BORDER='#dde1e8'; TEXT='#2c2c3a'
    TEXT_DIM='#6b7280'; TEXT_MUTED='#9ca3af'
    FONT="font-family:system-ui,-apple-system,'Segoe UI',Roboto,sans-serif"
    MONO="font-family:'SF Mono',SFMono-Regular,Consolas,monospace"
    C_OK=('#16a34a','#f0fdf4','#15803d'); C_WARN=('#d97706','#fffbeb','#92400e')
    C_ERR=('#dc2626','#fef2f2','#991b1b'); C_INFO=('#2563eb','#eff6ff','#1e40af')
    C_STEP=('#7c3aed','#f5f3ff','#5b21b6')
    def _card(i): return f'<div style="background:{BG_CARD};border:1px solid {BORDER};border-radius:16px;padding:2px;margin:14px 0"><div style="background:{BG_INNER};border-radius:14px;padding:24px">{i}</div></div>'
    def _header(ic,t,s=""):
        sh=f'<div style="{FONT};font-size:12.5px;color:{TEXT_DIM};margin-top:2px">{s}</div>' if s else ''
        return f'<div style="display:flex;align-items:center;gap:12px;margin-bottom:20px;padding-bottom:14px;border-bottom:2px solid {BORDER}"><div style="width:44px;height:44px;display:flex;align-items:center;justify-content:center;background:linear-gradient(135deg,#667eea,#764ba2);border-radius:12px;font-size:22px">{ic}</div><div><div style="{FONT};font-size:20px;font-weight:700;color:{TEXT}">{t}</div>{sh}</div></div>'
    def _row(m,c,ic=""):
        b,gb,tc=c; return f'<div style="display:flex;align-items:center;gap:10px;padding:10px 14px;margin:6px 0;border-radius:10px;border-left:3px solid {b};background:{gb};color:{tc};{MONO};font-size:13.5px">{ic} {m}</div>'
    def ce_show(icon,title,sub=""): display(HTML(_card(_header(icon,title,sub))))
    def ce_ok(m,ic="✅"): display(HTML(_row(m,C_OK,ic)))
    def ce_warn(m,ic="⚠️"): display(HTML(_row(m,C_WARN,ic)))
    def ce_err(m,ic="❌"): display(HTML(_row(m,C_ERR,ic)))
    def ce_info(m,ic="ℹ️"): display(HTML(_row(m,C_INFO,ic)))
    def ce_step(m,ic="🔹"): display(HTML(_row(m,C_STEP,ic)))

# ── Validate URL ──
if not url or not url.strip():
    ce_err("Please enter a URL in the field above")
    raise SystemExit("No URL provided")

url = url.strip()
if not url.startswith(('http://', 'https://')):
    url = 'https://' + url

parsed = urlparse(url)
if not parsed.netloc:
    ce_err(f"Invalid URL: {url}")
    raise SystemExit("Invalid URL")

ce_show("🕷️", "Scraping", url)

# ── Build driver ──
ce_step("Launching headless Chromium...")

chrome_opts = Options()
chrome_opts.binary_location = "/usr/bin/chromium-browser"
for arg in [
    "--headless=new", "--no-sandbox", "--disable-dev-shm-usage",
    "--disable-gpu", "--disable-extensions", "--disable-infobars",
    "--window-size=1920,1080",
    "--user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/125.0.0.0 Safari/537.36",
]:
    chrome_opts.add_argument(arg)

service = Service(executable_path="/usr/lib/chromium/chromedriver")
driver = webdriver.Chrome(service=service, options=chrome_opts)
driver.set_page_load_timeout(timeout)

ce_ok("Chromium launched")

# ── Navigate ──
ce_step(f"Loading page (timeout: {timeout}s)...")
start_time = time.time()

try:
    driver.get(url)

    # Wait for readyState
    WebDriverWait(driver, timeout).until(
        lambda d: d.execute_script("return document.readyState") == "complete"
    )
    ce_ok(f"Page loaded ({round(time.time() - start_time, 1)}s)")

    # Extra settle time
    if wait_after_load > 0:
        ce_step(f"Waiting {wait_after_load}s for JS to settle...")
        time.sleep(wait_after_load)

    # Auto-scroll
    scroll_count = 0
    if auto_scroll:
        ce_step("Scrolling to load lazy content...")
        last_h = driver.execute_script("return document.body.scrollHeight")
        for _ in range(50):
            driver.execute_script("window.scrollTo(0, document.body.scrollHeight)")
            time.sleep(0.8)
            new_h = driver.execute_script("return document.body.scrollHeight")
            scroll_count += 1
            if new_h == last_h:
                break
            last_h = new_h
        driver.execute_script("window.scrollTo(0, 0)")
        ce_ok(f"Scrolled {scroll_count} times")

    # Custom JS
    if custom_js and custom_js.strip():
        ce_step("Executing custom JavaScript...")
        try:
            driver.execute_script(custom_js)
            time.sleep(1)
            ce_ok("Custom JS executed")
        except Exception as js_err:
            ce_warn(f"Custom JS error: {str(js_err)[:100]}")

    # ── Capture HTML ──
    ce_step("Capturing rendered HTML...")
    html = driver.page_source
    final_url = driver.current_url
    title = driver.title
    load_time = round(time.time() - start_time, 2)

    ce_ok(f"Captured {len(html):,} characters ({len(html.encode('utf-8')) / 1024:.1f} KB)")

except Exception as e:
    ce_err(f"Failed to load page: {str(e)[:200]}")
    driver.quit()
    raise

finally:
    driver.quit()

# ── Parse metadata ──
ce_step("Extracting metadata...")
soup = BeautifulSoup(html, "html.parser")

meta_tags = {}
for tag in soup.find_all("meta"):
    name = tag.get("name") or tag.get("property") or tag.get("http-equiv")
    content = tag.get("content")
    if name and content:
        meta_tags[name] = content

links = []
for a in soup.find_all("a", href=True):
    href = a["href"]
    if not href.startswith(('http://', 'https://', 'mailto:', 'tel:', '#')):
        href = urljoin(final_url, href)
    links.append({"text": a.get_text(strip=True)[:120], "url": href})

images = []
for img in soup.find_all("img"):
    src = img.get("src", "")
    if src and not src.startswith(('data:', 'http://', 'https://')):
        src = urljoin(final_url, src)
    images.append({"src": src, "alt": img.get("alt", "")})

scripts = len(soup.find_all("script"))
stylesheets = len(soup.find_all("link", rel="stylesheet"))

content_hash = hashlib.sha256(html.encode("utf-8")).hexdigest()[:16]

ce_ok(f"Found {len(links)} links, {len(images)} images, {scripts} scripts, {stylesheets} stylesheets")

# ── Stats ──
def _stat_card(icon, label, value, color="#667eea"):
    return f'<div style="text-align:center;padding:16px 12px;background:{BG_INNER};border:1px solid {BORDER};border-radius:12px;min-width:120px;flex:1"><div style="font-size:24px;margin-bottom:6px">{icon}</div><div style="{FONT};font-size:22px;font-weight:700;color:{color}">{value}</div><div style="{FONT};font-size:11px;color:{TEXT_MUTED};margin-top:2px">{label}</div></div>'

stats_html = _stat_card("📄", "Characters", f"{len(html):,}")
stats_html += _stat_card("📦", "Size", f"{len(html.encode('utf-8')) / 1024:.1f} KB")
stats_html += _stat_card("🔗", "Links", str(len(links)))
stats_html += _stat_card("🖼️", "Images", str(len(images)))
stats_html += _stat_card("⏱️", "Load Time", f"{load_time}s")
stats_html += _stat_card("📜", "Scrolls", str(scroll_count))
display(HTML(f'<div style="display:flex;gap:10px;flex-wrap:wrap;margin:14px 0">{stats_html}</div>'))

# ── Redirect notice ──
if final_url != url:
    ce_info(f"Redirected: {url} → {final_url}", "↪️")

# ── Title & Description ──
if title:
    ce_ok(f"Title: {title}", "🏷️")

desc = meta_tags.get("og:description") or meta_tags.get("description") or meta_tags.get("twitter:description")
if desc:
    ce_info(f"Description: {desc[:200]}")

# ── Save HTML ──
if save_html_file:
    domain = parsed.netloc.replace("www.", "")
    safe_name = "".join(c if c.isalnum() or c in ('-', '.') else '_' for c in domain)
    ts = datetime.now().strftime("%Y%m%d_%H%M%S")
    filename = f"{safe_name}_{ts}.html"
    filepath = os.path.join(OUTPUT_DIR, filename)
    with open(filepath, 'w', encoding='utf-8') as f:
        f.write(html)
    ce_ok(f"Saved: {filepath}", "💾")

    # Also save metadata JSON
    meta_path = filepath.replace('.html', '_meta.json')
    meta_data = {
        "url": url,
        "final_url": final_url,
        "title": title,
        "char_count": len(html),
        "file_size_bytes": len(html.encode('utf-8')),
        "link_count": len(links),
        "image_count": len(images),
        "script_count": scripts,
        "stylesheet_count": stylesheets,
        "scroll_count": scroll_count,
        "load_time_s": load_time,
        "content_hash": content_hash,
        "meta_tags": meta_tags,
        "scraped_at": datetime.now(timezone.utc).isoformat(),
    }
    with open(meta_path, 'w', encoding='utf-8') as f:
        json.dump(meta_data, f, indent=2, ensure_ascii=False)
    ce_ok(f"Metadata: {meta_path}", "📋")

# ── HTML Preview ──
if show_preview:
    max_chars = 5000
    truncated = len(html) > max_chars
    preview = html[:max_chars]
    if truncated:
        preview += f"\n\n<!-- ... truncated ({len(html):,} total chars) ... -->"

    import html as html_lib
    escaped = html_lib.escape(preview)
    display(HTML(
        f'<div style="margin:14px 0">'
        f'<div style="{FONT};font-size:13px;font-weight:600;color:{TEXT};margin-bottom:8px">'
        f'📄 HTML Preview {"(truncated)" if truncated else "(full)"}</div>'
        f'<pre style="background:#1e1e2e;color:#d4d4d4;padding:16px;border-radius:12px;'
        f'overflow-x:auto;font-size:12px;max-height:400px;overflow-y:auto">'
        f'<code>{escaped}</code></pre></div>'
    ))

# ── Links Table ──
if show_links and links:
    max_rows = 30
    rows = ""
    import html as html_mod
    for i, lnk in enumerate(links[:max_rows]):
        txt = html_mod.escape(lnk["text"] or "(no text)")
        href = html_mod.escape(lnk["url"])
        rows += (
            f'<tr style="border-bottom:1px solid {BORDER}">'
            f'<td style="padding:6px 10px;font-size:12px;color:{TEXT_DIM}">{i+1}</td>'
            f'<td style="padding:6px 10px;font-size:12px;color:{TEXT};max-width:200px;'
            f'overflow:hidden;text-overflow:ellipsis;white-space:nowrap">{txt}</td>'
            f'<td style="padding:6px 10px;font-size:11px;{MONO};color:#2563eb;'
            f'max-width:350px;overflow:hidden;text-overflow:ellipsis;white-space:nowrap">'
            f'<a href="{href}" target="_blank">{href}</a></td></tr>'
        )
    extra = f"<div style='{FONT};font-size:11px;color:{TEXT_MUTED};margin-top:6px'>... and {len(links) - max_rows} more links</div>" if len(links) > max_rows else ""
    display(HTML(
        f'<div style="margin:14px 0">'
        f'<div style="{FONT};font-size:13px;font-weight:600;color:{TEXT};margin-bottom:8px">'
        f'🔗 Links ({len(links)} found)</div>'
        f'<div style="overflow-x:auto">'
        f'<table style="width:100%;border-collapse:collapse;border:1px solid {BORDER};border-radius:8px">'
        f'<thead><tr style="background:{BG_CARD}">'
        f'<th style="padding:8px 10px;text-align:left;font-size:11px;color:{TEXT_MUTED}">#</th>'
        f'<th style="padding:8px 10px;text-align:left;font-size:11px;color:{TEXT_MUTED}">Text</th>'
        f'<th style="padding:8px 10px;text-align:left;font-size:11px;color:{TEXT_MUTED}">URL</th>'
        f'</tr></thead><tbody>{rows}</tbody></table></div>{extra}</div>'
    ))

# ── Meta Tags ──
if meta_tags:
    important_keys = ['og:title', 'og:description', 'og:image', 'og:url', 'og:type',
                      'twitter:card', 'twitter:title', 'twitter:description',
                      'description', 'keywords', 'author', 'viewport', 'robots']
    shown = {k: v for k, v in meta_tags.items() if k in important_keys}
    if shown:
        meta_rows = ""
        import html as h3
        for k, v in shown.items():
            meta_rows += (f'<tr style="border-bottom:1px solid {BORDER}">'
                f'<td style="padding:6px 10px;font-size:12px;{MONO};color:#7c3aed;font-weight:600">{h3.escape(k)}</td>'
                f'<td style="padding:6px 10px;font-size:12px;color:{TEXT};max-width:400px;'
                f'overflow:hidden;text-overflow:ellipsis;white-space:nowrap">{h3.escape(v[:200])}</td></tr>')
        display(HTML(
            f'<div style="margin:14px 0">'
            f'<div style="{FONT};font-size:13px;font-weight:600;color:{TEXT};margin-bottom:8px">'
            f'🏷️ Key Meta Tags</div>'
            f'<table style="width:100%;border-collapse:collapse;border:1px solid {BORDER};border-radius:8px">'
            f'<thead><tr style="background:{BG_CARD}">'
            f'<th style="padding:8px 10px;text-align:left;font-size:11px;color:{TEXT_MUTED}">Tag</th>'
            f'<th style="padding:8px 10px;text-align:left;font-size:11px;color:{TEXT_MUTED}">Value</th>'
            f'</tr></thead><tbody>{meta_rows}</tbody></table></div>'
        ))

ce_show("✅", "Done!", f"Scraped {title or url} in {load_time}s")
ce_info("Run this cell again with a different URL to scrape more pages", "🔄")
```

```python
#@title 💾 Step 3 — Download & Export { display-mode: "form" }

#@markdown Download all scraped files as a ZIP archive.

#@markdown ---

import os, glob, zipfile
from datetime import datetime
from IPython.display import display, HTML

# ── UI helpers ──
try:
    ce_show
except NameError:
    BG_CARD='#f4f5f7'; BG_INNER='#ffffff'; BORDER='#dde1e8'; TEXT='#2c2c3a'
    TEXT_DIM='#6b7280'; TEXT_MUTED='#9ca3af'
    FONT="font-family:system-ui,-apple-system,'Segoe UI',Roboto,sans-serif"
    MONO="font-family:'SF Mono',SFMono-Regular,Consolas,monospace"
    C_OK=('#16a34a','#f0fdf4','#15803d'); C_WARN=('#d97706','#fffbeb','#92400e')
    C_ERR=('#dc2626','#fef2f2','#991b1b'); C_INFO=('#2563eb','#eff6ff','#1e40af')
    C_STEP=('#7c3aed','#f5f3ff','#5b21b6')
    def _card(i): return f'<div style="background:{BG_CARD};border:1px solid {BORDER};border-radius:16px;padding:2px;margin:14px 0"><div style="background:{BG_INNER};border-radius:14px;padding:24px">{i}</div></div>'
    def _header(ic,t,s=""):
        sh=f'<div style="{FONT};font-size:12.5px;color:{TEXT_DIM};margin-top:2px">{s}</div>' if s else ''
        return f'<div style="display:flex;align-items:center;gap:12px;margin-bottom:20px;padding-bottom:14px;border-bottom:2px solid {BORDER}"><div style="width:44px;height:44px;display:flex;align-items:center;justify-content:center;background:linear-gradient(135deg,#667eea,#764ba2);border-radius:12px;font-size:22px">{ic}</div><div><div style="{FONT};font-size:20px;font-weight:700;color:{TEXT}">{t}</div>{sh}</div></div>'
    def _row(m,c,ic=""):
        b,gb,tc=c; return f'<div style="display:flex;align-items:center;gap:10px;padding:10px 14px;margin:6px 0;border-radius:10px;border-left:3px solid {b};background:{gb};color:{tc};{MONO};font-size:13.5px">{ic} {m}</div>'
    def ce_show(icon,title,sub=""): display(HTML(_card(_header(icon,title,sub))))
    def ce_ok(m,ic="✅"): display(HTML(_row(m,C_OK,ic)))
    def ce_warn(m,ic="⚠️"): display(HTML(_row(m,C_WARN,ic)))
    def ce_err(m,ic="❌"): display(HTML(_row(m,C_ERR,ic)))
    def ce_info(m,ic="ℹ️"): display(HTML(_row(m,C_INFO,ic)))
    def ce_step(m,ic="🔹"): display(HTML(_row(m,C_STEP,ic)))

OUTPUT_DIR = "/content/webscrap_output"

ce_show("💾", "Export", "Download scraped files")

# ── Find files ──
html_files = sorted(glob.glob(os.path.join(OUTPUT_DIR, "*.html")))
json_files = sorted(glob.glob(os.path.join(OUTPUT_DIR, "*.json")))
all_files = html_files + json_files

if not all_files:
    ce_warn("No files found. Run Step 2 first to scrape a website.")
else:
    ce_info(f"Found {len(html_files)} HTML file(s) and {len(json_files)} metadata file(s)")

    # ── Create ZIP ──
    ts = datetime.now().strftime("%Y%m%d_%H%M%S")
    zip_name = f"webscrap_export_{ts}.zip"
    zip_path = os.path.join("/content", zip_name)

    with zipfile.ZipFile(zip_path, 'w', zipfile.ZIP_DEFLATED) as zf:
        for fpath in all_files:
            zf.write(fpath, os.path.basename(fpath))

    zip_size = os.path.getsize(zip_path) / 1024
    ce_ok(f"Created {zip_name} ({zip_size:.1f} KB)")

    # ── List contents ──
    for f in all_files:
        size = os.path.getsize(f) / 1024
        ce_info(f"  📄 {os.path.basename(f)} ({size:.1f} KB)")

    # ── Download ──
    try:
        from google.colab import files as colab_files
        ce_step("Triggering download...")
        colab_files.download(zip_path)
        ce_ok(f"Downloaded {zip_name}")
    except ImportError:
        ce_info(f"Not on Colab. File saved at: {zip_path}")
    except Exception as e:
        ce_warn(f"Download failed: {str(e)[:100]}")
        ce_info(f"File is at: {zip_path}")

ce_show("✅", "All Done!", "Your scraped data is ready")
```
